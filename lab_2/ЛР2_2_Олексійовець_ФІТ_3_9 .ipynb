## Завдання 13

Обчисліть відсоток пасажирів, які вижили, окремо для 1-го, 2-го та 3-го класів (`Pclass`). Визначте, який клас мав найвищий рівень виживання.

```python
class1_surv_rate = df[(df['Pclass'] == 1) & (df['Survived'] == 1)].shape[0] / df[df['Pclass'] == 1].shape[0]
class2_surv_rate = df[(df['Pclass'] == 2) & (df['Survived'] == 1)].shape[0] / df[df['Pclass'] == 2].shape[0]
class3_surv_rate = df[(df['Pclass'] == 3) & (df['Survived'] == 1)].shape[0] / df[df['Pclass'] == 3].shape[0]

print(f'Частка пасажирів 1-го класу, які вижили: {class1_surv_rate * 100:.2f}%')
print(f'Частка пасажирів 2-го класу, які вижили: {class2_surv_rate * 100:.2f}%')
print(f'Частка пасажирів 3-го класу, які вижили: {class3_surv_rate * 100:.2f}%')
```

Найвищий рівень виживання мали пасажири першого класу.

## Завдання 14

Обчисліть і виведіть середній вік пасажирів, які вижили, та пасажирів, які не вижили.

```python
survived_mean_age = df[df['Survived'] == 1]['Age'].mean()
not_survived_mean_age = df[df['Survived'] == 0]['Age'].mean()

print(f'Середній вік пасажирів, які вижили: {survived_mean_age:.2f}')
print(f'Середній вік пасажирів, які не вижили: {not_survived_mean_age:.2f}')
```

## Завдання 15

Розподіліть пасажирів на групи за рівнями тарифів (Fare) і обчисліть рівень виживання для кожної групи.

```python
df['Fare_Category'] = pd.qcut(df['Fare'], 6, labels=False)


for i in range(6):
    surv_rate = df[(df['Fare_Category'] == i) & (df['Survived'] == 1)].shape[0] / df[df['Fare_Category'] == i].shape[0]
    print(f'Частка пасажирів у тарифній групі {i}, які вижили: {surv_rate * 100:.2f}%')
```

## Завдання 16

Аналіз класу та тарифу: Визначте середній тариф (Fare) для кожного класу (Pclass).

```python
for i in range(1, 4):
    median_fare = df[df['Pclass'] == i]['Fare'].median()
    print(f'Медіанний тариф для класу {i}: {median_fare:.2f}')
```

## Завдання 17

Як взаємоповʼязані середній вік з виживанням?

```python
df_survived = df[df['Survived'] == 1][['Age']]
df_survived['Status'] = 'Вижили'

df_not_survived = df[df['Survived'] == 0][['Age']]
df_not_survived['Status'] = 'Не вижили'

df_combined = pd.concat([df_survived, df_not_survived])


sns.histplot(data=df_combined, x='Age', hue='Status', bins=20)
```

## Завдання 19

```python
mtx = df.drop('Survived', axis=1).corr(numeric_only=True).abs()


fig, ax = plt.subplots(figsize=(6, 6))

sns.heatmap(
    mtx,
    cmap='viridis',
    annot=True,
    fmt=".2f",
    linewidths=0.5,
    mask=np.tril(np.ones_like(mtx, dtype=bool)),
    square=True,
    cbar=False,
    ax=ax
)

plt.title("Матриця попарної кореляції ознак", fontsize=14)
plt.tight_layout()
plt.show()
```
